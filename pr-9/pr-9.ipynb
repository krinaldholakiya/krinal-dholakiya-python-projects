```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

class DataAnalyzer:
    def __init__(self):
        self.df = None

    def load_dataset(self):
        print("\n===== Load Dataset =====")
        load_choice = input("Enter The Path Of The Dataset (CSV File): ")
        try:
            self.df = pd.read_csv(load_choice)
            print("Dataset Loaded Successfully!")
        except Exception as e:
            print(f"Error loading file: {e}")

    def explore_data(self):
        if self.df is None:
            print("Please load a dataset first!")
            return
        while True:
            print("\n========== Explore Data ==========")
            print("1. Display the first 5 rows")
            print("2. Display the last 5 rows")
            print("3. Display column names")
            print("4. Display data types")
            print("5. Display basic info")
            print("6. Exit")
            case2_choice = int(input("Enter your choice: "))
            match case2_choice:
                case 1: print(self.df.head(5))
                case 2: print(self.df.tail(5))
                case 3: print(self.df.columns)
                case 4: print(self.df.dtypes)
                case 5: print(self.df.info())
                case 6: break

    def dataframe_operations(self):
        if self.df is None:
            print("Please load a dataset first!")
            return
        while True:
            print("\n========== Dataframe Operations ==========")
            print("1. Display All Product Names")
            print("2. Display All Category Names")
            print("3. Display Maximum And Minimum discount")
            print("4. Display Highest And lowest customer Rating")
            print("5. Exit")
            case3_choice = int(input("Enter Your Choice: "))
            match case3_choice:
                case 1:
                    print("---------- Product Name ----------")
                    print(self.df["Product"].unique())
                case 2:
                    print("---------- Category Name ----------")
                    print(self.df["Category"].unique())
                case 3:
                    print(f"----- Maximum Discount -----\n{self.df['DiscountPct'].max()}")
                    print(f"----- Minimum Discount -----\n{self.df['DiscountPct'].min()}")
                case 4:
                    print(f"----- Highest Rating -----\n{self.df['CustomerRating'].max()}")
                    print(f"----- Lowest Rating -----\n{self.df['CustomerRating'].min()}")
                case 5:
                    break

    def handle_missing_data(self):
        if self.df is None:
            print("Please load a dataset first!")
            return
        while True:
            print("\n========== Handle Missing Data ==========")
            print("1. Display rows with missing values")
            print("2. Fill missing values with mean or '-' ")
            print("3. Drop rows with missing values")
            print("4. Exit")
            case4_choice = int(input("Enter your choice: "))
            match case4_choice:
                case 1:
                    print(self.df.isna())
                case 2:
                    self.df.fillna({'UnitPrice': self.df['UnitPrice'].mean()}, inplace=True)
                    self.df.fillna({'Product': "-"}, inplace=True)
                    self.df.fillna({'Region': "-"}, inplace=True)
                    self.df.fillna({'Quantity': self.df['Quantity'].mean()}, inplace=True) 
                    self.df.fillna({'DiscountPct': self.df['DiscountPct'].mean()}, inplace=True)
                    self.df.fillna({'CustomerRating': self.df['CustomerRating'].mean()}, inplace=True) 
                    print("---------- Task Completed! ----------") 
                case 3:
                    self.df.dropna(inplace=True) 
                    print("---------- Task Completed! ----------")
                case 4:
                    break

    def data_visualization(self):
        if self.df is None:
            print("Please load a dataset first!")
            return
        while True:
            print("\n===== Data Visualization =====")
            print("1. Bar Plot")
            print("2. Line Plot")
            print("3. Scatter Plot")
            print("4. Pie Chart")
            print("5. Histogram")
            print("6. Stack Plot")
            print("7. Exit")
            case6_choice = int(input("Enter your choice: "))
            
            plot_generated = False
            match case6_choice:
                case 1:
                    plt.figure(figsize=(10,5))
                    sns.barplot(x="Product", y="Sales", data=self.df, errorbar=None)
                    x_label = input("Enter xlable: ")
                    y_label = input("Enter ylable: ")
                    plt.xlabel(x_label)
                    plt.ylabel(y_label)
                    plt.title("Product Sales")
                    plot_generated = True
                    if plot_generated:
                            save = input("Do You Want To Save Graph?(yes or no): ")
                    if save.lower() == "yes":
                            plt.savefig("barplot.png", dpi=300, bbox_inches='tight')
                            print("Your Graph Saved Successfully!")
                    plt.show()
                case 2:
                    plt.figure(figsize=(10,5))
                    sns.lineplot(data=self.df, x="Product", y="Sales", marker="o")
                    plt.title("Product Sales")
                    plot_generated = True
                    if plot_generated:
                            save = input("Do You Want To Save Graph?(yes or no): ")
                    if save.lower() == "yes":
                            plt.savefig("lineplot.png", dpi=300, bbox_inches='tight')
                            print("Your Graph Saved Successfully!")
                    plt.show()
                case 3:
                    plt.figure(figsize=(8,5))
                    sns.scatterplot(data=self.df, x="Product", y="Sales", hue="Region")
                    plt.title("Product Sales")
                    plot_generated = True
                    if plot_generated:
                            save = input("Do You Want To Save Graph?(yes or no): ")
                    if save.lower() == "yes":
                            plt.savefig("scatterplot.png", dpi=300, bbox_inches='tight')
                            print("Your Graph Saved Successfully!")
                    plt.show()
                case 4:
                    plt.figure(figsize=(8, 8))
                    df_grouped = self.df.groupby('Product')['Sales'].sum().reset_index()
                    plt.pie(df_grouped['Sales'], labels=df_grouped['Product'], autopct='%1.1f%%', startangle=140, shadow=True)                
                    plt.title("Product wise Sales Distribution")
                    plot_generated = True
                    if plot_generated:
                            save = input("Do You Want To Save Graph?(yes or no): ")
                    if save.lower() == "yes":
                            plt.savefig("piechart.png", dpi=300, bbox_inches='tight')
                            print("Your Graph Saved Successfully!")
                    plt.show()
                case 5:
                    plt.figure(figsize=(10, 5))
                    sns.histplot(data=self.df, x="Sales", kde=True, bins=20, color="purple")
                    plt.xlabel("Sales Range")
                    plt.ylabel("Frequency")
                    plt.title("Distribution of Sales")
                    plot_generated = True
                    if plot_generated:
                            save = input("Do You Want To Save Graph?(yes or no): ")
                    if save.lower() == "yes":
                            plt.savefig("histogram.png", dpi=300, bbox_inches='tight')
                            print("Your Graph Saved Successfully!")
                    plt.show()
                case 6:
                    plt.figure(figsize=(10, 5))
                    self.df['Date'] = pd.to_datetime(self.df['Date'])
                    self.df['Month'] = self.df['Date'].dt.to_period('M').astype(str)
                    df_pivot = self.df.groupby(['Month', 'Product'])['Sales'].sum().unstack().fillna(0)
                    plt.stackplot(df_pivot.index, df_pivot.values.T, labels=df_pivot.columns)
                    plt.xlabel("Month")
                    plt.ylabel("Total Sales")
                    plt.title("Product wise Sales")
                    plt.xticks(rotation=45)
                    plt.legend(loc="upper left")
                    plot_generated = True
                    if plot_generated:
                            save = input("Do You Want To Save Graph?(yes or no): ")
                    if save.lower() == "yes":
                            plt.savefig("stackplot.png", dpi=300, bbox_inches='tight')
                            print("Your Graph Saved Successfully!")
                    plt.show()

                case 7:
                    break

    def run(self):
       
        while True:
            print("\nPlease Select an Option:")
            print("1. Load Dataset")
            print("2. Explore Data")
            print("3. Perform Dataframe Operations")
            print("4. Handle Missing Data")
            print("5. Generate Descriptive Statistics")
            print("6. Data Visualization")
            print("7. Exit")
            
            try:
                choice = int(input("Enter Your Choice: "))
            except ValueError:
                print("Please enter a valid number.")
                continue
                
            match choice:
                case 1: self.load_dataset()
                case 2: self.explore_data()
                case 3: self.dataframe_operations()
                case 4: self.handle_missing_data()
                case 5:
                    if self.df is None:
                        print("Please load a dataset first!")
                    else:
                        print(self.df.describe())
                case 6: self.data_visualization()
                case 7:
                    print("-------------------- Thank You For Using Visualizer --------------------")
                    break

analyzer = DataAnalyzer()
analyzer.run()
```
